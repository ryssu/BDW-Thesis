Getting the Average Confidence score

```python
import json
import numpy as np


# Load predictions
json_path = "C:/Users/Carlos/Desktop/Thesis-Runs/Val/yolov5m_final_fog_val/best_predictions.json"

with open(json_path, 'r') as f:
    predictions = json.load(f)

# Extract all confidence scores
confidences = [pred['score'] for pred in predictions]

# Calculate statistics
print("="*60)
print("FOGGY WEATHER - Confidence Analysis")
print("="*60)
print(f"Total detections: {len(confidences)}")
print(f"Average confidence: {np.mean(confidences):.4f}")
print(f"Median confidence: {np.median(confidences):.4f}")
print(f"Std deviation: {np.std(confidences):.4f}")
print(f"Min confidence: {min(confidences):.4f}")
print(f"Max confidence: {max(confidences):.4f}")
print(f"25th percentile: {np.percentile(confidences, 25):.4f}")
print(f"75th percentile: {np.percentile(confidences, 75):.4f}")
```
